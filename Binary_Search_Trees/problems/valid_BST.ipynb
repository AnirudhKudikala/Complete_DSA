```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.data = val
        self.left = left
        self.right = right

class Solution:
    def validBST(self, root):
        self.inOrderTraversalStack=[]
        def inOrderTraversal(node):
            if not node:
                return
            inOrderTraversal(node.left)
            self.inOrderTraversalStack.append(node.data)
            inOrderTraversal(node.right)
            return self.inOrderTraversalStack
        st=inOrderTraversal(root)
        for i in range(1,len(st)):
            if st[i]<=st[i-1]:
                return False
        return True


#--------------------------------------------------------------------------------------------
        # def dfs(node, low, high):
        #     if not node:
        #         return True

        #     if not low<node.data<high:
        #         return False

        #     return dfs(node.left, low, node.data) and dfs(node.right, node.data, high)
        
        # return dfs(root, float("-inf"), float("inf"))
#--------------------------------------------------------------------------------------------
        # self.prev=None
        # def dfs(node):
        #     if not node:
        #         return True
        #     if not dfs(node.left):
        #         return False
        #     if self.prev and self.prev>=node.data:
        #         return False
        #     self.prev=node.data
        #     return dfs(node.right)
        # return dfs(root)

# Example usage
if __name__ == "__main__":
    # Constructing the tree: [3, 1, 4, None, 2]
    root = TreeNode(3)
    root.left = TreeNode(1)
    root.left.right = TreeNode(2)
    root.right = TreeNode(4)
    
    solution = Solution()
    result = solution.validBST(root)
    
    print(result)
```
