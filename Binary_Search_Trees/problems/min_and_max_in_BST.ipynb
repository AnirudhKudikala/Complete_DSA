```python
class TreeNode:
    def __init__(self, value=0):
        self.data = value
        self.left = None
        self.right = None

def findMin(root):
    if not root:
        return None
    node=root
    while node.left:
        node=node.left
    return node.data

def findMax(root):
    if not root:
        return None
    node=root
    while node.right:
        node=node.right
    return node.data


newBST = TreeNode(5)
newBST.left = TreeNode(3)
newBST.right = TreeNode(6)
newBST.left.left = TreeNode(2)
newBST.left.right = TreeNode(4)

print(findMin(newBST))
print(findMax(newBST))
```
