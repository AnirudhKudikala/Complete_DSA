```python
from collections import deque


class Solution:
    # Function to perform DFS traversal
    def dfs(self, start, graph):
        visited=set()
        result=[]
        def dfs_helper(node):
            visited.add(node)
            result.append(node)

            # Traverse all neighbours
            
            for neighbour in graph[node]:
                if neighbour not in visited:
                    dfs_helper(neighbour)
            # traverse the entire graph, even if it's disconnected
            # 0 -- 1

            # 2 -- 3

            # 4
            # for neighbour in range(len(graph)):
            #     if neighbour not in visited:
            #         dfs_helper(neighbour)
        dfs_helper(start)
        return result

    # Function to perform BFS traversal
    def bfs(self, start, graph):
        visited=set()
        result=[]
        visited.add(start)
        q=deque([start])
        while q:
            node=q.popleft()
            result.append(node)
            for neighbour in graph[node]:
                if neighbour not in visited:
                    visited.add(neighbour)
                    q.append(neighbour)
        return result


def main():
    # Number of vertices
    V = 5

    # Adjacency list
    adj = [[] for _ in range(V)]
    adj[0] = [1, 2]
    adj[1] = [0, 3]
    adj[2] = [0, 4]
    adj[3] = [1]
    adj[4] = [2]

    # Create object
    sol = Solution()

    # Run DFS from node 0
    resultDFS = sol.dfs(0, adj)

    resultBFS = sol.bfs(0, adj)

    # Print traversal
    print("DFS: ", " ".join(map(str, resultDFS)))
    print("BFS: ", " ".join(map(str, resultBFS)))


if __name__ == "__main__":
    main()
```
