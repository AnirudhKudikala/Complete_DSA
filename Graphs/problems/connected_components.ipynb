```python
from collections import deque

class Solution:
    # Function to count connected components in an undirected graph
    def countComponents(self, v, edges):
        adj_list=[[] for _ in range(v)]
        for u, v in edges:
            adj_list[u].append(v)
            adj_list[v].append(u)
        components=0
        visited=set()
        for i in range(v):
            if i not in visited:
                components+=1
                q=deque()
                q.append(i)
                visited.add(i)
                while q:
                    node = q.popleft()
                    for neighbour in adj_list[node]:
                        if neighbour not in visited:
                            visited.add(neighbour)
                            q.append(neighbour)
        return components

# Number of vertices
V = 5

# List of undirected edges
# edges = [[0, 1], [1, 2], [3, 4]]
edges=[[0,1],[0,2],[0,3],[1,3], [2,3]]

# Create solution object
sol = Solution()

# Print the number of connected components
print("Number of Connected Components:", sol.countComponents(V, edges))
```
