```python

# Node class for the binary tree
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def minDepthRecursive(self, root):
        if not root:
            return 0
        if not root.left and not root.right:
            return 1
        if not root.left:
            return 1+self.minDepthRecursive(root.right)
        if not root.right:
            return 1+self.minDepthRecursive(root.left)
        return 1+min(self.minDepthRecursive(root.left), self.minDepthRecursive(self.right))


# Creating a sample binary tree
root = Node(1)
root.right = Node(3)
root.right.right = Node(4)
root.right.right.right = Node(6)
root.right.right.right.right = Node(7)
root.right.right.right.right.right = Node(8)

solution = Solution()
minDepthRecursive = solution.minDepthRecursive(root)

print("Minimum depth of the binary tree recursive:", minDepthRecursive)
                           
```
