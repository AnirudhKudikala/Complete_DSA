```python
from collections import deque
class Node:
    def __init__(self, val):
        self.data = val
        self.left = None
        self.right = None

class Solution:
    def isLeafNode(self, node):
        return not node.left and not node.right

    def addLeftBoundary(self, node, res):
        curr=node.left
        while curr:
            if not self.isLeafNode(curr):
                res.append(curr.data)
            if curr.left:
                curr=curr.left
            else:
                curr=curr.right

    def addLeafBoundary(self, node, res):
        if self.isLeafNode(node):
            res.append(node.data)
        if node.left:
            self.addLeafBoundary(node.left, res)
        if node.right:
            self.addLeafBoundary(node.right, res)

    def addRightBoundary(self, node, res):
        curr=node.right
        q=deque()
        while curr:
            if not self.isLeafNode(curr):
                q.appendleft(curr.data)
            if curr.right:
                curr=curr.right
            else:
                curr=curr.left
        res+=list(q)

    def printBoundary(self, root):
        res=[]
        if not root:
            return res
        if not self.isLeafNode(root):
            res.append(root.data)
        self.addLeftBoundary(root, res)
        self.addLeafBoundary(root,res)
        self.addRightBoundary(root, res)
        return res

# Helper function to
# print the result
def printResult(result):
    for val in result:
        print(val, end=" ")
    print()

# Creating a sample binary tree
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)
root.right.right = Node(7)
root.right.right.left = Node(8)
root.right.right.right = Node(9)

solution = Solution()

# Get the boundary traversal
result = solution.printBoundary(root)

# Print the result
print("Boundary Traversal:", end=" ")
printResult(result)
```
