```python
class Solution:
    # Function to check if the given array is a min-heap
    def isMinHeap(self, nums):
        n=len(nums)
        for i in range(n//2):
            left=2*i+1
            if left < n and nums[i]>=nums[left]:
                return False
            right=2*i+1
            if right < n and nums[i]>=nums[right]:
                return False
        return True

# Driver code
if __name__ == "__main__":
    obj = Solution()
    nums = [10, 20, 30, 21, 23]

    # Output result
    output = obj.isMinHeap(nums)
    print("true" if output else "false")
```
