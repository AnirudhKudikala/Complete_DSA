```python
import heapq

def merge_k_sorted_arrays(lists):
    heap=[]
    res=[]

    for arr_ind, arr in enumerate(lists):
        if arr:
            heapq.heappush(heap, (arr[0], arr_ind, 0))

    while heap:
        value, arr_indx, ele_ind = heapq.heappop(heap)
        res.append(value)

        next_indx = ele_ind+1
        if next_indx<len(lists[arr_indx]):
            heapq.heappush(heap, (lists[arr_indx][next_indx], arr_indx, next_indx))
    return res

lists = [[1,4,5],[1,3,4],[2,6]]
print(merge_k_sorted_arrays(lists))
# Output: [1,1,2,3,4,4,5,6]
```
