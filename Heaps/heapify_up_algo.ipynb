```python
class MinHeap:
    def __init__(self):
        self.heap = []

    def printHeap(self):
        return self.heap

    def insert(self, value):
        self.heap.append(value)
        self.heapify_up(len(self.heap) - 1)

    def heapify_up(self, index):
        while index > 0:
            parent = (index - 1) // 2

            if self.heap[parent] <= self.heap[index]:
                break

            self.heap[parent], self.heap[index] = (
                self.heap[index],
                self.heap[parent],
            )

            index = parent

minheap=MinHeap()
minheap.insert(20)
minheap.insert(10)
minheap.insert(30)
print(minheap.printHeap())
```
