Insertion Sort works like how most people arrange a hand of playing cards.

You pick one card at a time and insert it into its correct position among the cards you've already sorted.

Time complexity
| Case                   | Complexity |
| ---------------------- | ---------- |
| Best (already sorted)  |   O(n)     |
| Average                |   O(n²)    |
| Worst (reverse sorted) |   O(n²)    |

Space complexity
O(1)

Stable: It preserves the original relative order of duplicate elements

```python
def insertion_sort(arr):
    n = len(arr)

    # Start from the second element because
    # the first element is already considered sorted.
    for i in range(1, n):

        # Element to be inserted
        key = arr[i]

        # Compare with previous elements
        j = i - 1

        # Shift all elements greater than key
        # one position to the right.
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        # Insert key into its correct position.
        arr[j + 1] = key

    return arr


arr = [5, 3, 8, 4, 2]
print(insertion_sort(arr))
```
