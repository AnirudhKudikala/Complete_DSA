```python
# Quick Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

print(quick_sort([3, 6, 8, 10, 1, 2, 1]))

# Time Complexity: O(n log n)
# Space Complexity: O(log n) in the best case, O(n) in the worst case because of the recursion stack.

# Quick Sort is not a stable sort, meaning that equal elements in the input array may appear in 
# different positions in the output array.

# Quick Sort is an in-place sort, meaning that it does not require additional array for the output.
```
