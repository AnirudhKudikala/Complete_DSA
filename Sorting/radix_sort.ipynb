```python
# Radix Sort
def counting_sort(arr, exp):
    n = len(arr)
    output = [0] * n
    count = [0] * 10
    for i in range(n):
        index = arr[i] // exp
        count[index % 10] += 1 
    for i in range(1, 10):
        count[i] += count[i - 1]
    i = n - 1
    while i >= 0:
        index = arr[i] // exp
        output[count[index % 10] - 1] = arr[i]
        count[index % 10] -= 1
        i -= 1  
    for i in range(n):
        arr[i] = output[i]     
    return arr

def radix_sort(arr):
    max_num = max(arr)
    exp = 1
    while max_num // exp > 0:
        counting_sort(arr, exp)
        exp *= 10
    return arr

print(radix_sort([170, 45, 75, 90, 802, 24, 2, 66]))

# Time Complexity: O(n * k) where k is the number of digits in the maximum number.
# Space Complexity: O(n + k)

# Radix Sort is a stable sort, meaning that equal elements in the input array will appear in the 
# output array in the same order as they did in the input.

# Radix Sort is not an in-place sort, meaning that it requires additional array for the output.
```
