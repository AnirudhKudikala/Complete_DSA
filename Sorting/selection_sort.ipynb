Selection Sort is a straightforward comparison-based algorithm that sorts an array by repeatedly isolating the smallest element from the unsorted section and swapping it to its correct position

Time complexity
| Case                   | Complexity |
| ---------------------- | ---------- |
| Best (already sorted)  |   O(n²)    |
| Average                |   O(n²)    |
| Worst                  |   O(n²)    |

Space complexity
O(1)

Unstable - Default implementation can swap equal elements out of order.

```python
def selection_sort(arr):
    n = len(arr)
    
    for i in range(n - 1):
        # Assume the current position holds the minimum value
        min_idx = i
        
        # Scan the remaining unsorted portion
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
                
        # Swap the found minimum with the first unsorted element
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        
    return arr

numbers = [64, 25, 12, 22, 11]
sorted_numbers = selection_sort(numbers)
print("Sorted array:", sorted_numbers)
```
