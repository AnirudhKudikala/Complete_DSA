```python
# Merge Sort
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    # Recursively sort the left and right halves
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    # Merge the sorted halves
    return merge(left, right)

print(merge_sort([38, 27, 43, 3, 9, 82, 10]))

# Time Complexity: O(n log n)
# Space Complexity: O(n)

# Merge Sort is a stable sort, meaning that equal elements in the input array will appear in the 
# output array in the same order as they did in the input.

# Merge Sort is not an in-place sort, meaning that it requires additional array for the output.
```
