```python
# Definition for singly-linked list.
#You are given an array of integers nums and the head of a linked list. 
# Return the head of the modified linked list after removing all nodes 
# from the linked list that have a value that exists in nums.

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def modifiedList(self, nums, head):
        sett=set(nums)
        dummy=ListNode(-1)
        dummy.next=head
        prev=dummy
        curr=head

        while curr:
            if curr.val in sett:
                prev.next=curr.next
            else:
                prev=curr
            curr=curr.next
        return dummy.next

sol=Solution()
head=ListNode(1)
head.next=ListNode(2)
head.next.next=ListNode(1)
head.next.next.next=ListNode(2)
head.next.next.next.next=ListNode(1)
head.next.next.next.next.next=ListNode(2)
result_linked_list= sol.modifiedList([1],head)
# Convert the result_linked_list to a list
result_list=[]
while result_linked_list:
    result_list.append(result_linked_list.val)
    result_linked_list=result_linked_list.next
print(result_list)
```
