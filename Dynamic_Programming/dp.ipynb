Dynamic Programming can be described as storing answers to various sub-problems to be used later whenever required to solve the main problem.

The two common dynamic programming approaches are:

*Memoization*: Known as the "top-down" dynamic programming, usually the problem is solved in the direction of the main problem to the base cases.

*Tabulation*: Known as the "bottom-up" dynamic programming, usually the problem is solved in the direction of solving the base cases to the main problem

Note: The base case does not always mean smaller input. It depends upon the implementation of the algorithm.

Example: Fibonacci Series

```python
# Memoization
class Solution:
    # Function to calculate Fibonacci using memoization
    def fib(self, n, dp):
        # If base case return n
        if n <= 1:
            return n

        # If already computed, return stored value
        if dp[n] != -1:
            return dp[n]

        # Otherwise compute and store
        dp[n] = self.fib(n - 1, dp) + self.fib(n - 2, dp)
        return dp[n]

# Driver
if __name__ == "__main__":
    n = 6
    dp = [-1] * (n + 1)
    sol = Solution()
    print(sol.fib(n, dp))
```

```python
# Tabulation
class Solution:
    # Function to calculate Fibonacci using tabulation
    def fib(self, n):
        # If n is 0 or 1, return n
        if n <= 1:
            return n

        # Create dp array
        dp = [0] * (n + 1)

        # Initialize base cases
        dp[0] = 0
        dp[1] = 1

        # Fill dp array iteratively
        for i in range(2, n + 1):
            dp[i] = dp[i - 1] + dp[i - 2]

        # Return final answer
        return dp[n]

# Driver
if __name__ == "__main__":
    n = 6
    sol = Solution()
    print(sol.fib(n))
```

```python
# Space optimization
class Solution:
    def fib(self, n: int) -> int:
        # If n is 0 return 0
        if n == 0:
            return 0
        # If n is 1 return 1
        if n == 1:
            return 1

        # prev2 stores fib(n-2)
        prev2 = 0
        # prev stores fib(n-1)
        prev = 1

        # Loop from 2 to n
        for i in range(2, n + 1):
            # Calculate current fib
            curr = prev + prev2
            # Update prev2
            prev2 = prev
            # Update prev
            prev = curr

        # Return final answer
        return curr

def main():
    s = Solution()
    n = 6
    print(s.fib(n))

if __name__ == "__main__":
    main()
```
