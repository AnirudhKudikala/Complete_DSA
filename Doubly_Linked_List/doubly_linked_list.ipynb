```python
# Create Node
class Node:
    def __init__(self, value):
        self.value=value
        self.next=None
        self.prev=None
```

```python
def insert_at_end(head, node):
    curr=head
    while curr.next:
        curr=curr.next
    curr.next=node
    node.prev=curr
    return head
    
```

```python
def insert_at_first(head, node):
    node.next=head
    head.prev=node
    head=node
    return head
```

```python
def delete_head(head):
    if not head or not head.next:
        return None
    head=head.next
    head.prev=None
    return head
```
